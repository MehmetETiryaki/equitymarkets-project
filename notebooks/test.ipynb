```python
import statistics
import pandas as pd
import sys
sys.path.append("..")
import equity as eq
import equity_strats as eq_strats
import matplotlib.pyplot as plt
import math
```

```python
AAPL_options_csv = '../data/apple_options.csv'
AAPL_underlying_csv = '../data/apple_stocks.csv'
AAPL_options = pd.read_csv(AAPL_options_csv, parse_dates = ["exdate", "date"])
AAPL_underlying = pd.read_csv(AAPL_underlying_csv, parse_dates = ["date"])
V_options_csv = '../data/visa_options.csv'
V_underlying_csv = '../data/visa_stocks.csv'
V_options = pd.read_csv(V_options_csv, parse_dates = ["exdate", "date"])
V_underlying = pd.read_csv(V_underlying_csv, parse_dates = ["date"])
CSCO_options_csv = '../data/cisco_options.csv'
CSCO_underlying_csv = '../data/cisco_stocks.csv'
CSCO_options = pd.read_csv(CSCO_options_csv, parse_dates = ["exdate", "date"])
CSCO_underlying = pd.read_csv(CSCO_underlying_csv, parse_dates = ["date"])
INTC_options_csv = '../data/intel_options.csv'
INTC_underlying_csv = '../data/intel_stocks.csv'
INTC_options = pd.read_csv(INTC_options_csv, parse_dates = ["exdate", "date"])
INTC_underlying = pd.read_csv(INTC_underlying_csv, parse_dates = ["date"])
MSFT_options_csv = '../data/msft_options.csv'
MSFT_underlying_csv = '../data/msft_stocks.csv'
MSFT_options = pd.read_csv(MSFT_options_csv, parse_dates = ["exdate", "date"])
MSFT_underlying = pd.read_csv(MSFT_underlying_csv, parse_dates = ["date"])
XLK_options_csv = '../data/XLK_options.csv'
XLK_underlying_csv = '../data/XLK_stocks.csv'
XLK_options = pd.read_csv(XLK_options_csv, parse_dates = ["exdate", "date"])
XLK_underlying = pd.read_csv(XLK_underlying_csv, parse_dates = ["date"])
risk_free = pd.read_csv("../data/DGS10.csv", parse_dates = ["DATE"])
```

```python
options_dict = {"AAPL": AAPL_options_csv, "MSFT": MSFT_options_csv, "CSCO": CSCO_options_csv, "V": V_options_csv, "INTC": INTC_options_csv}
stocks_dict = {"AAPL": AAPL_underlying_csv, "MSFT": MSFT_underlying_csv, "CSCO": CSCO_underlying_csv, "V": V_underlying_csv, "INTC": INTC_underlying_csv}
options_df_dict = {"AAPL": AAPL_options, "MSFT": MSFT_options, "CSCO": CSCO_options, "V": V_options, "INTC": INTC_options}
stocks_df_dict = {"AAPL": AAPL_underlying, "MSFT": MSFT_underlying, "CSCO": CSCO_underlying, "V": V_underlying, "INTC": INTC_underlying}
weightings = {"AAPL": 0.34, "MSFT": 0.36, "CSCO": 0.11, "V": 0.096, "INTC": 0.094}

options_weightings = eq.calculate_allocation_premium_neutral(10, XLK_options_csv, XLK_underlying_csv, options_dict, stocks_dict, pd.Timestamp(2019,5,31), 90, 0.05, weightings)

daily_returns = eq_strats.backtest_dispersion(options_df_dict, stocks_df_dict, XLK_options, XLK_underlying, risk_free, options_weightings, pd.Timestamp(2019, 5, 31), 90, 10)
```

```python

# Assuming daily_returns is a pandas Series with a datetime index
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(daily_returns.index, daily_returns.values, marker='o', linestyle='-', color='b')  # Plot daily returns
plt.title('Daily Returns Over Time')  # Set title
plt.xlabel('Date')  # Set x-axis label
plt.ylabel('Returns')  # Set y-axis label
plt.grid(True)  # Show grid
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.tight_layout()  # Adjust layout to make room for the rotated x-axis labels
plt.show()  # Display the plot
```

```python
daily_returns_V = eq_strats.backtest_short_straddle_with_premium_change(V_options, V_underlying, risk_free, pd.Timestamp(2022, 6, 1), 1, 180, True)
series_V = daily_returns_V.set_index('date')['daily pnl']
daily_returns_AAPL = eq_strats.backtest_short_straddle_with_premium_change(AAPL_options, AAPL_underlying, risk_free, pd.Timestamp(2022, 6, 1), 1, 180, True)
series_AAPL = daily_returns_AAPL.set_index('date')['daily pnl']
daily_returns_MSFT = eq_strats.backtest_short_straddle_with_premium_change(MSFT_options, MSFT_underlying, risk_free, pd.Timestamp(2022, 6, 1), 1, 180, True)
series_MSFT = daily_returns_MSFT.set_index('date')['daily pnl']
daily_returns_CSCO = eq_strats.backtest_short_straddle_with_premium_change(CSCO_options, CSCO_underlying, risk_free, pd.Timestamp(2022, 6, 1), 1, 180, True)
series_CSCO = daily_returns_CSCO.set_index('date')['daily pnl']
daily_returns_INTC = eq_strats.backtest_short_straddle_with_premium_change(INTC_options, INTC_underlying, risk_free, pd.Timestamp(2022, 6, 1), 1, 180, True)
series_INTC = daily_returns_INTC.set_index('date')['daily pnl']
```

```python
print(len(series_V))
print(len(series_AAPL))
print(len(series_MSFT))
print(len(series_CSCO))
print(len(series_INTC))

daily_returns_total = series_V + series_AAPL + series_MSFT + series_INTC
```

```python

# Assuming daily_returns is a pandas Series with a datetime index
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(daily_returns_total.index, daily_returns_total.values, marker='o', linestyle='-', color='b')  # Plot daily returns
plt.title('Daily Returns Over Time')  # Set title
plt.xlabel('Date')  # Set x-axis label
plt.ylabel('Returns')  # Set y-axis label
plt.grid(True)  # Show grid
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.tight_layout()  # Adjust layout to make room for the rotated x-axis labels
plt.show()  # Display the plot
```

```python
print(statistics.mean(daily_returns_total.values))
print(statistics.stdev(daily_returns_total.values))
```
