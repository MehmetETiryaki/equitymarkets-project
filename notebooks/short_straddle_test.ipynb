```python
import statistics
import pandas as pd
import sys
sys.path.append("..")
import equity as eq
import equity_strats as eq_strats
import matplotlib.pyplot as plt
import math
```

```python
AAPL_options_csv = '../data/AAPL_options.csv'
AAPL_underlying_csv = '../data/AAPL_underlying.csv'
AAPL_options = pd.read_csv(AAPL_options_csv, parse_dates = ["exdate", "date"])
AAPL_underlying = pd.read_csv(AAPL_underlying_csv, parse_dates = ["date"])
AVGO_options_csv = '../data/avgo_options.csv'
AVGO_underlying_csv = '../data/avgo_underlying.csv'
AVGO_options = pd.read_csv(AVGO_options_csv, parse_dates = ["exdate", "date"])
AVGO_underlying = pd.read_csv(AVGO_underlying_csv, parse_dates = ["date"])
NVDA_options_csv = '../data/nvda_options.csv'
NVDA_underlying_csv = '../data/nvda_underlying.csv'
NVDA_options = pd.read_csv(NVDA_options_csv, parse_dates = ["exdate", "date"])
NVDA_underlying = pd.read_csv(NVDA_underlying_csv, parse_dates = ["date"])
MSFT_options_csv = '../data/MSFT_options.csv'
MSFT_underlying_csv = '../data/MSFT_underlying.csv'
MSFT_options = pd.read_csv(MSFT_options_csv, parse_dates = ["exdate", "date"])
MSFT_underlying = pd.read_csv(MSFT_underlying_csv, parse_dates = ["date"])
VZ_options_csv = '../data/VZ_options.csv'
VZ_underlying_csv = '../data/VZ_underlying.csv'
VZ_options = pd.read_csv(VZ_options_csv, parse_dates = ["exdate", "date"])
VZ_underlying = pd.read_csv(VZ_underlying_csv, parse_dates = ["date"])
risk_free = pd.read_csv("../data/DGS10.csv", parse_dates = ["DATE"])
```

```python
daily_returns_AAPL = eq_strats.backtest_short_straddle_with_premium_change(AAPL_options, AAPL_underlying, risk_free, pd.Timestamp(2018, 1, 2), 1, 180, True)
series_AAPL = daily_returns_AAPL.set_index('date')['daily pnl']

```

```python
start_date = pd.Timestamp(2018, 1, 2)
series_AAPL = pd.Series()
for i in range(9):

    daily_returns_AAPL = eq_strats.backtest_short_straddle_with_premium_change(AAPL_options, AAPL_underlying, risk_free, start_date, 1, 180, True)
    series_AAPL = series_AAPL.add(daily_returns_AAPL.set_index('date')['daily pnl'], fill_value = 0)
    start_date = series_AAPL.index[-1]
```

```python
start_date = pd.Timestamp(2018, 1, 2)
series_MSFT = pd.Series()

for i in range(9):

    daily_returns_MSFT = eq_strats.backtest_short_straddle_with_premium_change(MSFT_options, MSFT_underlying, risk_free, start_date, 1, 180, True)
    series_MSFT = series_MSFT.add(daily_returns_MSFT.set_index('date')['daily pnl'], fill_value = 0)
    start_date = series_MSFT.index[-1]
```

```python

# Assuming daily_returns is a pandas Series with a datetime index
plt.figure(figsize=(10, 6))  # Set the figure size
plt.plot(series_MSFT.index, series_MSFT.values, marker='o', linestyle='-', color='b')  # Plot daily returns
plt.title('Daily Returns Over Time')  # Set title
plt.xlabel('Date')  # Set x-axis label
plt.ylabel('Returns')  # Set y-axis label
plt.grid(True)  # Show grid
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability
plt.tight_layout()  # Adjust layout to make room for the rotated x-axis labels
plt.show()  # Display the plot

print(statistics.stdev(series_MSFT.values))
print(statistics.mean(series_MSFT.values))
```
